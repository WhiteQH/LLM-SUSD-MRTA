```python
import csv
import random

# 设置环境尺寸（可修改为300或其他尺寸）
width = 100    # 环境宽度（米）
height = 100   # 环境高度（米）

# 配置生成的任务组数量及每组内的坐标数量
num_groups = 20         # 任务组数量
coords_per_group = 20  # 每组坐标个数

# 输出文件名称
filename = "task_coordinates.csv"

with open(filename, 'w', newline='', encoding='utf-8') as csvfile:
    writer = csv.writer(csvfile)
    # 写入表头
    writer.writerow(['Group', 'task_coordinates'])
    
    # 逐组生成随机坐标并写入CSV文件
    for group in range(1, num_groups + 1):
        # 生成一组坐标，每个坐标的x在[0, width]，y在[0, height]之间
        coordinates = [[round(random.uniform(0, width), 4), round(random.uniform(0, height), 4)] 
                       for _ in range(coords_per_group)]
        # 将坐标列表转换为字符串，格式与示例保持一致
        coord_str = str(coordinates)
        writer.writerow([group, coord_str])

print(f"生成的CSV文件已保存为：{filename}")
```
